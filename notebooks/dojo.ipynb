# Welcome to the AI Dojo

> **_dōjō_ (道場, Japanese) -  a place for immersive learning. Literally, a place to learn the path (_dō_).**

The AI Dojo is a place for learning about AI and Data Science. It is a place to learn about the tools, techniques, and best practices for building AI applications. 

Our approach to teaching is to provide a hands-on experience to get you started. For this we provide:
- introductory material
- code demos
- interactive coding exercises

## Paths

The generative AI courses build on the precursor courses **MLP** and **DLT**, which are available in [**Data Science Learning Paths**](https://github.com/point8/data-science-learning-paths).

### Machine Learning with Python [MLP]

A 2-day intensive course enabling the participants to build machine learning applications with Python and the scikit-learn framework. We work on typical machine learning cases using hands-on examples.

- **Level**: Intermediate
- **Duration**: 2 days
- **Prerequisites**: Python programming, basic data analysis with Python
- **Index notebook**: [📓 Start course](https://github.com/point8/data-science-learning-paths/blob/main/notebooks/index/mlp2-machine-learning-python-2day.ipynb) _(in Data Science Learning Paths)_


### Deep Learning with TensorFlow [DLT]

Get started with neural networks and deep learning, applying the TensorFlow and Keras libraries.

- **Level**: Intermediate
- **Duration**: 2 days
- **Prerequisites**: MLP
- **Index notebook**: [📓 Start course](https://github.com/point8/data-science-learning-paths/blob/main/notebooks/index/dlt2-intro-dl-tensorflow-2day.ipynb) _(in Data Science Learning Paths)_


### Generative AI & Large Language Models

Explore latest advances in generative AI and work with open large language models.

- **Level**: Advanced
- **Duration**: 2 days
- **Prerequisites**: DLT
- **Index notebook**: [📓 Start course](index/ai-genai-llm.ipynb)



### Developing Generative AI Assistants

On the path to your company's own AI bot — learn to build intelligent AI assistants using Large Language Models (LLMs) and Retrieval-Augmented Generation (RAG). From embeddings and transformer architecture to production-ready chatbots with tool use and agentic patterns — hands-on with both proprietary APIs and self-hosted open-source models.

- **Level**: Advanced
- **Duration**: 3 days
- **Prerequisites**: Python programming
- **Index notebook**: [📓 Start course](index/ai-genai-assistants.ipynb)
