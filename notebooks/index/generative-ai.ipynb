# Generative AI

In this course we focus on recent innovations in **Generative AI**, particularly **Generative Language Models**, the technology behind ChatGPT. You learn how language models work, how to choose and run them — through proprietary APIs or as self-hosted open models — and how to build intelligent AI assistants with Retrieval-Augmented Generation (RAG), structured outputs and tool use.
