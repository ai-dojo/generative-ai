# Developing Generative AI Assistants 

**On the path to your company's own AI bot**

In this 3-day course, you will learn to build intelligent AI assistants using Large Language Models (LLMs) and Retrieval-Augmented Generation (RAG). From understanding embeddings and transformer architecture to building production-ready chatbots with tool use and agentic patterns — you'll gain hands-on experience with both proprietary APIs and self-hosted open-source models.
