# Generative AI & Large Language Models

In this course we focus on recent innovations in **Generative AI**, particularly **Generative Language Models**, the technology behind the popular ChatGPT. We teach you to leverage the power of language models in your applications. Going beyond just using ChatGPT, we enable you to host your own open source chat models, build Python applications with them, and even create custom models for your applications.
