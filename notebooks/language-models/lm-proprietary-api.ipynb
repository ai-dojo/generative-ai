# Using Proprietary LLM APIs via Poe

Access multiple frontier language models through a single API using [Poe](https://poe.com/). 

Like many providers, the Poe API provides an **OpenAI-compatible endpoint**, which means we can use the `openai` Python library to access models from different providers—simply by changing the `base_url` to point to Poe's servers. This is a common pattern: many API providers implement the OpenAI API specification, making it easy to switch between them.

**Models covered:**
- **Claude Opus 4.5** (Anthropic)
- **GPT-5** (OpenAI)
- **Gemini 3** (Google)

```python
from openai import OpenAI
from ai_dojo import show

# Connect to Poe API (OpenAI-compatible endpoint)
client = OpenAI(
    api_key=input("Poe API key: "), 
    base_url="https://api.poe.com/v1"
)

# Same prompt for all models
PROMPT = "Write a haiku about artificial intelligence."
```

## Claude Opus 4.5 (Anthropic)

```python
response = client.chat.completions.create(
    model="Claude-4.5-Opus",
    messages=[{"role": "user", "content": PROMPT}],
    stream=True
)

show.stream(response)
```

## GPT-5 (OpenAI)

```python
response = client.chat.completions.create(
    model="GPT-5",
    messages=[{"role": "user", "content": PROMPT}],
    stream=True
)

show.stream(response)
```

## Gemini 3 (Google)

```python
response = client.chat.completions.create(
    model="Gemini-3-Pro",
    messages=[{"role": "user", "content": PROMPT}],
    stream=True
)

show.stream(response)
```
