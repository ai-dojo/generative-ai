## Use Cases

`llama_index` provides...

- an **abstraction layer for various LLMs** from different providers
- **data connectors** to ingest your existing data sources and data formats (APIs, PDFs, docs, SQL, etc.).
- **vector stores** for storing and retrieving your data
- **retrieval/query/chat interfaces** over your data: Feed in any LLM input prompt, get back retrieved context and knowledge-augmented output.

_(Adapted from the [LlamaIndex README](https://github.com/run-llama/llama_index).)_

## Connecting to LLMs from Different Providers

`llama_index` can serve as a common interface for including different LLM providers. It abstracts away some of the details of prompting LLMs.

### Open LLMs via Ollama

Working with `Ollama`? `llama_index` supports it via extension modules. 

```python
show.command("pip install llama-index-llms-ollama llama-index-embeddings-ollama")
```

```python
from llama_index.llms.ollama import Ollama
```

```python
from ai_dojo.lm import get_ollama_url

ollama_url = get_ollama_url()
ollama_url
```

```python
model = "olmo-3:7b"
# model = "mixtral:latest"

llm = Ollama(
    model=model,
    request_timeout=90.0,
    base_url=ollama_url,
)
```

**Completion**

```python
response = llm.complete("The meaning of life is...")
show.text(response.text)
```

**Chat**

```python
from llama_index.core.llms import ChatMessage, MessageRole
```

```python
messages = [
    ChatMessage(
        role=MessageRole.SYSTEM,
        content="You are a humorous assistant.",
    ),
    ChatMessage(
        role=MessageRole.USER, content="Tell me a joke about large language models"
    ),
]
```

```python
response = llm.chat(
    messages,
)
show.text(response.message.content)
```

```python
messages.append(response.message)
```

```python
messages.append(
    ChatMessage(
        role=MessageRole.USER,
        content="Explain why this was funny.",
    )
)
```

```python
response = llm.chat(
    messages,
)
show.text(response.message.content)
```

**Streaming responses**

```python
for response in llm.stream_complete("The meaning of life is"):
    print(response.delta, end="")
```

### Proprietary LLMs

This example needs an API key from [Anthropic](https://console.anthropic.com/). Set it as the environment variable `ANTHROPIC_API_KEY`, or enter it when prompted.

```python
# from llama_index.llms.openai import OpenAI
from llama_index.llms.anthropic import Anthropic

import os
from getpass import getpass
```

```python
llm = Anthropic(
    model="claude-3-opus-20240229",
    api_key=os.environ.get("ANTHROPIC_API_KEY") or getpass("Anthropic API key: "),
)
```

```python
response = llm.complete("Colorless green ideas...")
show.text(response.text)
```

```python
response = llm.chat(
    [
        ChatMessage(
            role=MessageRole.SYSTEM,
            content="You are a humorous assistant.",
        ),
        ChatMessage(
            role=MessageRole.USER,
            content="Tell me a joke about large language models",
        ),
    ],
)
show.text(response.message.content)
```
