# Ragadoc: RAG in Action

**An AI document assistant that answers questions about your PDFs with citations**

[Ragadoc](https://github.com/clstaudt/ragadoc) is an open-source Streamlit application that demonstrates RAG concepts in a practical, user-friendly way:

- 🤖 **AI Document Q&A** — Ask natural language questions about your PDFs
- 📍 **Citation Grounding** — Every answer includes specific citations from your document
- 🎯 **PDF Highlighting** — Citations are automatically highlighted in the original PDF
- 🔒 **Complete Privacy** — Uses only local AI models via Ollama

## Architecture

Ragadoc implements the RAG pattern we learned about:

```
PDF Upload → Text Extraction → Chunking → Vector Embeddings
                                               ↓
User Question → Semantic Search → Context Retrieval → AI Response
                                               ↓
                                    Citation Highlighting
```

**Tech Stack:**
- **Frontend**: Streamlit
- **AI Models**: Ollama (local LLMs)
- **Vector DB**: ChromaDB
- **PDF Processing**: PyMuPDF4LLM
- **Embeddings**: nomic-embed-text

## Quick Start

### Prerequisites

Make sure Ollama is running with the required models:

````python
import ollama

# Check available models
models = ollama.list()
model_names = [m.model for m in models.models]

required = ['nomic-embed-text', 'olmo3:7b']
for model in required:
    status = "✅" if any(model in name for name in model_names) else "❌ missing"
    print(f"{model}: {status}")
````

## How to Use

1. **Upload a PDF** — Drag and drop or browse for your document
2. **Select AI Model** — Choose from your locally installed Ollama models
3. **Start Chatting** — Ask questions about your document in natural language
4. **View Citations** — See highlighted text in the PDF that supports each answer
5. **Explore** — Continue the conversation to dive deeper into your document

## Model Recommendations

| Model Type | Model Name | Size | RAM Required | Use Case |

| **Chat** | olmo3:7b | ~4GB | 8GB | Good balance |
| **Chat** | olmo-3.1:32b | ~19GB | 32GB | Higher quality |
| **Chat** | phi3:mini | ~2.3GB | 4GB | Low-resource systems |

## Key RAG Concepts Demonstrated

Ragadoc shows several important RAG patterns:

1. **Document Chunking** — PDFs are split into manageable pieces for embedding
2. **Vector Storage** — ChromaDB stores embeddings for fast similarity search
3. **Semantic Retrieval** — Questions are matched to relevant chunks by meaning
4. **Context Augmentation** — Retrieved chunks are added to the LLM prompt
5. **Citation Tracking** — Source locations are preserved for transparency
