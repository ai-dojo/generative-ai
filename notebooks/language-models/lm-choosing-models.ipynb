### Measuring LLM Capability with Benchmarks

**MMLU** (Massive Multitask Language Understanding) is a widely-used benchmark for evaluating language models across 57 subjects including mathematics, history, law, medicine, and computer science. 

- **Human expert level:** ~89.8% (the dashed line in the charts)
- **Score range:** From ~40% (early models) to ~90% (frontier models)
- **What it measures:** Broad knowledge and reasoning across academic disciplines

The chart above shows how models from different organizations have progressed on MMLU over time. Note how both closed-source (proprietary) and open-weight models have steadily improved, with open models often catching up to proprietary leaders within 6-12 months.

### Proprietary, Open-Weight, and Open-Source Models

Not all "open" models are equally open. Understanding these distinctions matters for deployment, customization, and compliance:

| Category | Definition |
|----------|-----------|
| **Proprietary (Closed)** | Weights not released; access only via API. Provider controls all aspects. |
| **Open-Weight** | Model weights publicly available for download and local deployment. Training data/code often not released. |
| **Open-Source** | Weights, training data, training code, and evaluation all publicly available. Fully reproducible. |

## Key Decision Factors

### 1. 🔒 Data Privacy & Control

| Requirement | Recommendation |
|-------------|----------------|
| Sensitive data must stay on-premise | **Self-hosted open-source** |
| Data can go to cloud provider | Proprietary API or cloud-managed |
| Regulatory compliance (GDPR, HIPAA) | Check provider's data handling policies |

**Key question:** *Can your data leave your network?*

### 2. 💰 Cost Structure

| Model Type | Cost Model | Best For |
|------------|------------|----------|
| Proprietary API | Pay-per-token | Low/variable volume |
| Self-hosted (GPU) | Fixed infrastructure | High volume, predictable cost |
| Cloud-managed | Subscription + tokens | Enterprise with compliance needs |



### 3. ⚡ Model Size & Speed

| Size | Parameters | Use Case | Hardware |
|------|------------|----------|----------|
| **Tiny** | < 3B | Quick tests, edge devices | CPU |
| **Small** | 7-8B | Chat, simple tasks | Single GPU (8GB+) |
| **Medium** | 13-32B | Complex reasoning, coding | Single GPU (24GB+) |
| **Large** | 70B+ | Highest quality | Multi-GPU |

**Bigger isn't always better!** A well-tuned 7B model often beats a generic 70B model for specific tasks.

### 4. 🎯 Task Requirements

| Task | Recommended Approach |
|------|---------------------|
| General chat | Any instruction-tuned model |
| Code generation | Code-specialized models (CodeLlama, Qwen-Coder) |
| RAG / Q&A | Good instruction-following + embeddings model |
| Image understanding | Multimodal model (Qwen3.5, LLaVA) |
| Structured output | Models with good JSON mode support |
| Long documents | Models with large context windows |

## Exploring Available Models

```python
import ollama

# List models available on your Ollama instance
models = ollama.list()

for model in models.get('models', []):
    size_gb = model.get('size', 0) / 1e9
    print(f"{model['model']:30} {size_gb:.1f} GB")
```

```python
# Quick comparison: same prompt, different models
from IPython.display import Markdown, display
from ai_dojo import show

models = ["tinyllama",  "qwen3.5:4b", "olmo-3:7b",]

prompt = "Explain what a Large Language Model does to someone who has no technical background. One sentence only."

for model_name in models:
    try:
        display(Markdown(f"**{model_name}:**"))
        response = ollama.chat(
            model=model_name,   
            messages=[{"role": "user", "content": prompt}],
            stream=True
        )
        show.stream(response)
    except Exception as e:
        print(f"{model_name}: Not available ({e})")
```

## Summary: How to Choose

1. **Start with constraints:** Privacy requirements narrow your options first
2. **Match size to task:** Don't use a 70B model for simple classification
3. **Check the license:** Especially for commercial deployments
4. **Test on your data:** Benchmarks don't predict your specific use case
5. **Consider the ecosystem:** Support, documentation, community matter

**When in doubt:** Start with a smaller, open model. You can always upgrade later
