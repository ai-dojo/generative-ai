# Chat Models: From Text Completion to Instruction Following

We have seen that LLMs are trained to generate the next word/token given some input text.

But there is still a good way to go to turn such a model into a useful and pleasant chat model.

## Let's have a chat with a "raw" language model

```python
from ai_dojo import show
```

```python
import ollama
```

Some models in ollama are available in a `:text` and a `:instruct` version. The `:text` version is a "raw" LLM that has not been finetuned for chat and instruction following.

```python

model = "llama3:text"
ollama.pull(model)
```

```python
prompt = "Tell me a short story about a mouse that can travel back in time"

response = ollama.chat(
    model=model,
    messages=[{"role": "user", "content": prompt}],
    stream=True,
)
show.stream(response)
```

```python
prompt = "What is the meaning of the ≈ sign in mathematics?"

response = ollama.chat(
    model=model,
    messages=[{"role": "user", "content": prompt}],
    stream=True,
)
show.stream(response)
```

## And now with the same model, but finetuned to make it "chattable"

Models of the `:instruct` variant have been finetuned to make them more "chattable". (The "instruct" is implicit, if you just use "llama3", you will also get the instruct model.)

```python

model = "llama3:instruct"
ollama.pull(model)
```

```python
prompt = "Tell me a short story about a mouse that can travel back in time"

response = ollama.chat(
    model=model,
    messages=[{"role": "user", "content": prompt}],
    stream=True,
)
show.stream(response)
```

```python
prompt = "What is the meaning of the ≈ sign in mathematics?"

response = ollama.chat(
    model=model,
    messages=[{"role": "user", "content": prompt}],
    stream=True,
)
show.stream(response)
```

## How does this work?

1. Given different prompts, human labelers writes a suitable output text. The prompt-output pairs are used to fine-tune the model and teach it to _reply to_, rather than simply _continue_ the prompt.
2. Human labelers rank the outputs to different prompts. This data is used to train reward model to distinguish good from bad responses.
3. That reward model is used to further fine-tune the language model.
