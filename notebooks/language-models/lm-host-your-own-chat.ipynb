# Hosting your own AI Chat App - with Ollama and Open WebUI

Want your own open-source "ChatGPT"? In this notebook, we'll explore how to set up and host an AI chat app using  Open WebUI for the chat interface and Ollama for the AI backend. 

**Open WebUI** is an open-source clone of OpenAI's _ChatGPT_ web interface that allows you to chat with a great variety of language models from various providers. 

```python
from ai_dojo import show
```

```python
openwebui_url = "https://github.com/open-webui/open-webui"
show.github_repo(openwebui_url)
```

Open WebUI It has excellent integration with Ollama, so that you can be up and running without doing much configuration at all.

The only thing you need is the [docker-compose.yaml](https://github.com/open-webui/open-webui/blob/main/docker-compose.yaml) file from the Open WebUI repository. Then you can start both Open WebUI and Ollama by running `docker compose up`. 

Navigate to [http://localhost:3000](http://localhost:3000) and create a user account. (The user accounts are all local, there is no spam risk associated). 

After logging in, you are now able to interact with Ollama, download models and select them for a chat session:
